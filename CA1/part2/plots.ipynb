```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


df = pd.read_csv("results/experiment_results.csv")

df["TotalTime_µs"] = df["TotalTime_µs"].astype(float)
df["AvgLatency_µs"] = df["AvgLatency_µs"].astype(float)
df["StdLatency_µs"] = df["StdLatency_µs"].astype(float)
sns.set_theme(style="darkgrid")
```

```python
plt.figure(figsize=(10,6))
sns.lineplot(data=df, x="Goroutines", y="Throughput_tasks_per_sec", hue="GOMAXPROCS", style="Workload", markers=True, palette="tab10")
plt.title("Throughput vs Goroutines")
plt.ylabel("Throughput (tasks/sec)")
plt.xlabel("Number of Goroutines")
plt.tight_layout()
plt.legend()
plt.savefig("results/throughput_vs_goroutines.png")
```

```python
plt.figure(figsize=(10,6))
sns.lineplot(data=df, x="Goroutines", y="AvgLatency_µs", hue="GOMAXPROCS", style="Workload", markers=True, palette="tab10")
plt.title("Average Latency vs Goroutines")
plt.ylabel("Average Latency (µs)")
plt.xlabel("Number of Goroutines")
plt.tight_layout()
plt.legend()
plt.savefig("results/avg_latency_vs_goroutines.png")
```

```python
plt.figure(figsize=(10,6))
sns.lineplot(data=df, x="Goroutines", y="StdLatency_µs", hue="GOMAXPROCS", style="Workload", markers=True, palette="tab10")
plt.title("Std of Latency vs Goroutines")
plt.ylabel("Std (µs)")
plt.xlabel("Number of Goroutines")
plt.tight_layout()
plt.legend()
plt.savefig("results/std_latency_vs_goroutines.png")
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x="AvgLatency_µs", y="Throughput_tasks_per_sec", hue="Workload", style="GOMAXPROCS", s=50, palette="tab10")
plt.title("Throughput vs Average Latency") 
plt.xlabel("Avg Latency (µs)")
plt.ylabel("Throughput (tasks/sec)")
plt.tight_layout()
plt.legend()
plt.savefig("results/throughput_vs_latency.png")
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(data=df, x="Workload", y="AvgLatency_µs", hue="GOMAXPROCS", palette="tab10")
plt.title("Latency Boxplot")
plt.ylabel("Latency (µs)")
plt.tight_layout()
plt.legend()
plt.savefig("results/latency_boxplot.png")
```
